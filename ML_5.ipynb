Random Forest Classifier for Ensemble Learning

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)
```

```python
df = pd.read_csv("statement_dataset_realistic_200.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

print("\nFirst 5 Records:")
print(df.head())

print("\nClass Distribution:")
print(df["Label"].value_counts())
```

```python
X = df["Text"]
y = df["Label"]

```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("\nTraining records:", len(X_train))
print("Testing records :", len(X_test))
```

```python
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("\nTF-IDF Training Shape:", X_train_tfidf.shape)
print("TF-IDF Testing Shape :", X_test_tfidf.shape)

```

```python
rf_classifier = RandomForestClassifier(
    n_estimators=100,
    max_depth=8,
    min_samples_split=4,
    min_samples_leaf=2,
    random_state=42
)

rf_classifier.fit(
    X_train_tfidf,
    y_train
)

print("\nRandom Forest model trained successfully!")
```

```python
y_pred = rf_classifier.predict(X_test_tfidf)
```

```python
accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\n===================================")
print("RANDOM FOREST CLASSIFICATION")
print("===================================")

print("Accuracy:", accuracy)
print("Accuracy Percentage:", round(accuracy * 100, 2), "%")
```

```python

print("\nConfusion Matrix:")

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=["Negative", "Positive"]
)

print(cm)
```

```python
print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        labels=["Negative", "Positive"],
        zero_division=0
    )
)
```

```python
sample = [
    "I really enjoyed this product"
]

sample_tfidf = vectorizer.transform(sample)

prediction = rf_classifier.predict(sample_tfidf)

print("\n===================================")
print("SAMPLE PREDICTION")
print("===================================")

print("Sample Statement:", sample[0])
print("Predicted Class:", prediction[0])
```

```python
samples = [
    "The product was excellent and very useful",
    "The service was slow and disappointing",
    "I am happy with the quality",
    "The application was difficult to use",
    "The movie was enjoyable",
    "I regret buying this product"
]

samples_tfidf = vectorizer.transform(samples)
predictions = rf_classifier.predict(samples_tfidf)

print("\n===================================")
print("MULTIPLE SAMPLE PREDICTIONS")
print("===================================")

for text, pred in zip(samples, predictions):
    print("\nText:", text)
    print("Predicted Class:", pred)
```
