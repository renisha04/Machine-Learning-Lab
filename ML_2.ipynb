```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
df = pd.read_csv("student_performance_balanced_1000.csv")

print("Dataset:")
print(df.head())

print("\nDataset Shape:", df.shape)
```

```python
df["GradeCategory"] = pd.cut(
    df["FinalGrade"],
    bins=[-float("inf"), 50, 75, float("inf")],
    labels=["Low", "Medium", "High"]
)
```

```python
print("\nGrade Category Distribution:")
print(df["GradeCategory"].value_counts())
```

```python
X = df[
    [
        "AttendanceRate",
        "StudyHoursPerWeek",
        "PreviousGrade",
        "AssignmentScore"
    ]
]
```

```python
y = df["GradeCategory"]
```

```python
valid_target = y.notna()

X = X.loc[valid_target].copy()
y = y.loc[valid_target].copy()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Data Shape:", X_train.shape)
print("Testing Data Shape :", X_test.shape)
```

```python
svm_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),

    ("scaler", StandardScaler()),

    ("svm", SVC(
        kernel="rbf",
        C=1.0,
        gamma="scale",
        random_state=42
    ))
])
```

```python
svm_pipeline.fit(X_train, y_train)

print("\nSVM model trained successfully.")
```

```python
y_pred = svm_pipeline.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", accuracy)
```

```python
print("\nActual Values:")
print(y_test.values)
```

```python
print("\nPredicted Values:")
print(y_pred)

```

```python
print("\nClassification Report:\n")

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
print("\nConfusion Matrix:")

print(
    confusion_matrix(
        y_test,
        y_pred
    )
)
```

```python
print("\nActual Class Distribution:")
print(y_test.value_counts())
```

```python
print("\nPredicted Class Distribution:")
print(pd.Series(y_pred).value_counts())
```
