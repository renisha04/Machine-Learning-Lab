Text Categorization using SVM

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
```

Load the dataset

```python
data = pd.read_csv("statement_dataset.csv")

print("Dataset:")
print(data.head())

print("\nTotal number of records:", len(data))
```

Separate input and output

```python
X = data["Text"]
y = data["Label"]
```

Split dataset into training and testing data

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining records:", len(X_train))
print("Testing records :", len(X_test))
```

Convert text into numerical vectors

```python
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english"
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("\nTF-IDF Training Shape:", X_train_tfidf.shape)
print("TF-IDF Testing Shape :", X_test_tfidf.shape)
```

Create and train SVM model

```python
model = SVC(kernel="linear")

model.fit(X_train_tfidf, y_train)
```

Predict test data

```python
y_pred = model.predict(X_test_tfidf)
```

Calculate accuracy

```python
accuracy = accuracy_score(y_test, y_pred)

print("\nActual Labels    :", list(y_test))
print("Predicted Labels :", list(y_pred))

print("\nAccuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")
```

Classification report

```python
print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

Test with new text

```python
new_text = [
    "I really enjoyed this excellent product"
]

# Convert new text into TF-IDF vector
new_text_tfidf = vectorizer.transform(new_text)

# Predict category
prediction = model.predict(new_text_tfidf)

print("\nNew Text:", new_text[0])
print("Predicted Class:", prediction[0])
```
