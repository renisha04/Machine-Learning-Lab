```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
data = pd.read_csv("statement_dataset_realistic_200.csv")

print("Dataset loaded successfully!")

print("\nDataset:")
print(data.head())

print("\nTotal number of records:", len(data))

print("\nClass Distribution:")
print(data["Label"].value_counts())
```

```python
X = data["Text"]
y = data["Label"]

print("Input column: Text")
print("Output column: Label")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("\nTraining records:", len(X_train))
print("Testing records :", len(X_test))
```

```python
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("\nTF-IDF Training Shape:", X_train_tfidf.shape)
print("TF-IDF Testing Shape :", X_test_tfidf.shape)
```

```python
model = SVC(
    kernel="linear",
    C=0.5
)

model.fit(X_train_tfidf, y_train)

print("\nSVM model trained successfully!")
```

```python
y_pred = model.predict(X_test_tfidf)

print("\nActual Labels:")
print(list(y_test))

print("\nPredicted Labels:")
print(list(y_pred))
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("\n===================================")
print("ACCURACY")
print("===================================")

print("Accuracy:", accuracy)
print("Accuracy Percentage:", round(accuracy * 100, 2), "%")
```

```python
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=["Negative", "Positive"]
)

print("\n===================================")
print("CONFUSION MATRIX")
print("===================================")

print("              Predicted")
print("             Negative Positive")
print("--------------------------------")

print("Actual Negative", cm[0])
print("Actual Positive", cm[1])
```

```python
print("\n===================================")
print("CLASSIFICATION REPORT")
print("===================================")

print(
    classification_report(
        y_test,
        y_pred,
        labels=["Negative", "Positive"],
        zero_division=0
    )
)
```

```python
new_text = [
    "The product was useful and I am satisfied with my purchase"
]

new_text_tfidf = vectorizer.transform(new_text)

prediction = model.predict(new_text_tfidf)

print("\n===================================")
print("NEW TEXT PREDICTION")
print("===================================")

print("New Text:", new_text[0])
print("Predicted Class:", prediction[0])
```

```python
new_texts = [
    "I am very happy with the quality of this product",
    "The service was slow and disappointing",
    "The application is easy and convenient to use",
    "I regret buying this product because it stopped working",
    "The movie was enjoyable and entertaining",
    "The customer support was unhelpful and rude"
]

new_texts_tfidf = vectorizer.transform(new_texts)

predictions = model.predict(new_texts_tfidf)

print("\n===================================")
print("MULTIPLE TEXT PREDICTIONS")
print("===================================")

for text, prediction in zip(new_texts, predictions):
    print("\nText:", text)
    print("Predicted Class:", prediction)
```
